```python
# Create a dataloader 
enc_data_loader = DataLoader(folder="feature_encoded_ds")
fs_data_loader = DataLoader(folder=r"feature_sel_data\numeric")

# Load the numeric dataset
numeric_ds = fs_data_loader.collect(filename="num_feature_ds_v1.csv")

# Load the 2 encoded dataset
processor_brand_df = enc_data_loader.collect(filename="Processor_name_encoded_ds.csv")
weight_df = enc_data_loader.collect(filename="Weight_encoded_ds.csv")
os_df = enc_data_loader.collect(filename="Os_encoded_ds.csv")
```

## Data Transformation

```python
# Combine 3 dataset & exclude the number of irrelevant features 
# Processor Brand name: Select the features containing that are Intel (Core) and AMD ("Ryzen") 
processor_brand_names = [brand_name for brand_name in processor_brand_df.columns if ("Core" in brand_name) or ("Ryzen" in brand_name)]
sel_processor_brand_df = processor_brand_df[processor_brand_names]

# Remove the categroical weight feature from the weight df
# weight_df = weight_df.drop(columns=["Weight"]) 

# Extract price feature 
price_feature = numeric_ds[["Price"]]

# Combine the selected processor brand df & weight df and numeric df 
new_feature_ds = pd.concat([numeric_ds, sel_processor_brand_df, weight_df], axis=1)
new_feature_ds = new_feature_ds.drop(columns=["Price"], axis=True)
new_feature_ds = pd.concat([new_feature_ds, price_feature], axis=1)
new_feature_ds
```

```python
# Add OS feature into the new dataframe
sel_os_features = os_df[["Windows","Mac"]]
new_feature_ds = pd.concat([new_feature_ds, sel_os_features], axis=1)
new_feature_ds
```
